## 1. IMPORT NEEDED LIBRARY

```python
import pandas as pd
```

## 2. READ FILES

```python
uv_c = pd.read_csv('../data/processed/uv_cleaned.csv')
core_c = pd.read_csv('../data/processed/core_cleaned.csv')
outer_c = pd.read_csv('../data/processed/outer_cleaned.csv')

uv_b_c = pd.read_csv('../data/processed/uv_batch_cleaned.csv')
core_b_c = pd.read_csv('../data/processed/core_batch_cleaned.csv')
outer_b_c = pd.read_csv('../data/processed/outer_batch_cleaned.csv')

mix_c = pd.read_csv('../data/processed/mix_cleaned.csv')

result_c = pd.read_csv('../data/processed/result_cleaned.csv')
```

## 3. MERGE MIX WITH RESULT

```python
result_c.head(5)
```

```python
mix_c.head(5)
```

These two data frames will be merged via 'AXF Number', so verification needs to be performed.

```python
print(result_c['Batch number'].unique())
```

```python
print(mix_c['AXF Number'].unique())
```

There are a few experiments in 'result_cleaned' file with special markings ('AXF0045 (large)' and 'AXF0045 - 7 days') and some experiments with typographical errors where an extra space is present in the name. <br>
<br>
These experiments will be processed to reduce the number of unique values. <br>
<br>
File 'mix_cleaned' file shows no unusual signs.

```python
result_c['Batch number'] = result_c['Batch number'].str.strip()
print(result_c['Batch number'].unique())
```

It can be seen that experiment 'AXF0045' has records in two different special cases.

```python
result_c[result_c['Batch number'].str.contains('AXF0045')]
```

The head's experiments were always quality-checked at 24 hours, 48 ​​hours, and 7 days. <br>
However, the dataset consistently presented information on quality-check results within 24 hours and 48 hours. <br>
Therefore, the experiment line 'AXF0045 - 7 days' was removed.<br>
<br>
The experiment name 'AXF0045 (large)' will be changed to 'AXF0045'

```python
result_c = result_c.drop(result_c[(result_c['Batch number'] == 'AXF0045 - 7 days')].index)

result_c.loc[0, 'Batch number'] = 'AXF0045'
```

To join two data frames, we need to rename the 'Batch number' column in 'result_c' to 'AXF Number' first.

```python
result_c = result_c.rename(columns={'Batch number': 'AXF Number'})
merge_1 = mix_c.merge(result_c,how='left', on=['AXF Number'])
merge_1.info()
print('-'*50)
merge_1.head(5)
```

```python
merge_1.columns
```

'Date of experiment', 'Aims & Hypothesis' will be droped because they are not helpful in the analysis. <br>
'UV formulation_y' and 'Core formulation' will be dropped because they conflict with columns that share the same name and data as other columns in the data frame.<br>
'QC Pass/Fail?' will be dropped because the data in this column is considered based on the old condition (rule 35,50,75).



```python
merge_1 = merge_1.drop(columns=['Date of experiment', 
                                'Aims & Hypothesis',
                                'UV formulation_y',
                                'Core formulation ',
                                'QC Pass/Fail?'])
merge_1.rename(columns = {'UV formulation_x': 'UV formulation'}, inplace = True)
merge_1.info()
print('-'*50)
merge_1.head(5)
```

```python
merge_1.isna().sum().to_frame()
```

There are still 47 rows with missing data in '% dry after 24h in humidity' and '% dry after 2 days in humidity chamber'.

```python
merge_1 = merge_1.dropna()
merge_1.isna().sum().to_frame()
```

```python
merge_1.info()
print('-'*50)
merge_1.head(5)
```

The most important information in batch files is the formula used to create that batch and the resulting quality (Vicosity).<br>
Except for the Outer, the 'mix' file only contains information about the Batch but not the Outer Formulation. <br>
<br>
'merge_1' will be paired with 'outer_b' to retrieve the associated recipe information.<br>
The results will then be linked to each formulation file.

## 4. MERGE 'MERGE_1' AND 'OUTER_B'

```python
outer_b_c.columns
```

```python
outer_b_c['Batch Number '] = outer_b_c['Batch Number '].str.strip()
outer_b_c['Formulation'] = outer_b_c['Formulation'].str.strip()
```

```python
print(outer_b_c['Batch Number '].unique())
```

```python
print(outer_b_c['Formulation'].unique())
```

```python
outer_b_c.rename(columns = {'Batch Number ': 'Outer used'}, inplace = True)
```

The 'Formulation' column needs to be renamed to 'Outer formulation' to distinguish it from other formulations.

```python
merge_2.rename(columns = {'Formulation' : 'Outer formulation'}, inplace = True)
```

## 5. MERGE 'MERGE_2' AND FORMULATION FILES

```python
merge_2.columns
```

```python
uv_c.columns
```

```python
core_c.columns
```

```python
outer_c.columns
```

Notice that the first 'core_c' and 'outer_c' have the name column 'Water'; these need to be changed to differentiate them.

```python
core_c.rename(columns = {'Water ': 'Water_core'}, inplace = True)
outer_c.rename(columns = {'Water': 'Water_outer'}, inplace = True)
```

The 'Total' column in both 'uv_c' and 'core_c' is not necessary for analyzing the relationship with the object to be optimized. They will be dropped.

```python
uv_c = uv_c.drop(columns='Total')
core_c = core_c.drop(columns='Total')
```

The 'Sample' column of the formulation files needs to be renamed correctly according to the formulation for merging with 'merge_2'.

```python
uv_c.rename(columns = {'Sample': 'UV formulation'}, inplace = True)
core_c.rename(columns = {'Sample': 'Core Formulation'}, inplace = True)
outer_c.rename(columns = {'Sample': 'Outer formulation'}, inplace = True)
```

Merging files

```python
merge_3 = merge_2.merge(uv_c,how = 'left', on=['UV formulation'])
merge_4 = merge_3.merge(core_c,how = 'left', on=['Core Formulation'])
final = merge_4.merge(outer_c,how ='left', on=['Outer formulation'])
final.info()
print('-'*50)
final
```

```python
for col in final.columns:
    try:
        final[col] = final[col].astype('float64')
    except:
        pass
```
